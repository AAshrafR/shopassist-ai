# ShopAssist AI: Exploratory Data Analysis & Data Preprocessing

This notebook demonstrates:
1. Loading the **Bitext Customer Support LLM Chatbot Dataset**.
2. Data inspection, cleaning, and normalization.
3. Exploratory Data Analysis (EDA) on categories, intents, and text length distributions.
4. Knowledge base chunk generation and policy integration for RAG.
5. Generating the benchmark evaluation set.

```python
import pandas as pd
import numpy as np
import json
from pathlib import Path

RAW_CSV = Path('../data/raw/bitext_customer_support.csv')
df = pd.read_csv(RAW_CSV)
print('Dataset shape:', df.shape)
df.head()
```

## Category & Intent Distribution

```python
print('Unique Categories:', df['category'].nunique())
print(df['category'].value_counts())

print('\nUnique Intents:', df['intent'].nunique())
df['intent'].value_counts()
```

## Text Length Statistics

```python
df['instruction_length'] = df['instruction'].str.len()
df['response_length'] = df['response'].str.len()
print('Instruction length summary:')
print(df['instruction_length'].describe())
print('\nResponse length summary:')
print(df['response_length'].describe())
```

## Knowledge Base & Grounded Chunks

```python
KB_FILE = Path('../data/processed/knowledge_base_chunks.json')
with open(KB_FILE, 'r') as f:
    kb = json.load(f)
print(f'Total chunks in Knowledge Base: {len(kb)}')
print('Sample chunk:\n', json.dumps(kb[0], indent=2))
```
